```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Adaptive Ticket Auction Simulation")
```

```python
bids = {
    "Fan A": 80,
    "Fan B": 60,
    "Scalper A": 75
}

sorted_bids = sorted(bids.items(), key=lambda x: x[1], reverse=True)

winner = sorted_bids[0][0]
payment = sorted_bids[1][1]

print("Winner:", winner)
print("Payment:", payment)
```
